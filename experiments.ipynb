```python
from qiskit.quantum_info import Statevector
import numpy as np

from data import QuantumEnsemble, QuantumDataSource
```

## Quantum Data Examples

We will create a simple double class data source consisting of two ensembles of single-qubit quantum states.

```python
zero = Statevector.from_label("0")
one  = Statevector.from_label("1")
plus = (zero + one) / np.sqrt(2)
minus = (zero - one) / np.sqrt(2)

# Create a data source with only pure classes

# Class 0: |0>
ensemble_zero = QuantumEnsemble(
    states=[zero],
    probs=[1.0]
)
# Class 1: |1>
ensemble_one = QuantumEnsemble(
    states=[one],
    probs=[1.0]
)

data_source_pure = QuantumDataSource(
    ensembles=[ensemble_zero, ensemble_one]
)


# Create a data source with non-pure classes

# Class 0: |0> and |1>
ensemble_z = QuantumEnsemble(
    states=[zero, one],
    probs=[0.5, 0.5]
)
# Class 1: |+> and |->
ensemble_x = QuantumEnsemble(
    states=[plus, minus],
    probs=[0.5, 0.5]
)

data_source_non_pure = QuantumDataSource(
    ensembles=[ensemble_z, ensemble_x]
)
```

```python
data_source_non_pure.sample_class(0)
```
